### Objectives

- Build a FAISS vector index from training data as a knowledge base (KB)
- Query the index for relevant context and use a Cross-Encoder to re-rank results
- Run A/B tests comparing zero-shot inference with and without RAG context
- Explore context window limits and adversarial RAG dangers

### Key Concepts

| Concept | Description |
|---------|------------|
| **Bi-Encoder (FAISS)** | Embeds query and documents separately; fast but may miss nuance |
| **Cross-Encoder** | Passes query + document jointly through transformer; slow but accurate |
| **RAG** | Retrieval-Augmented Generation: inject retrieved context into model prompt |
| **Adversarial RAG** | Feeding incorrect context ΓÇö demonstrates LLM's over-trust in context |

## Setup

```python
!pip install faiss-cpu sentence-transformers --quiet

import pandas as pd
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer, CrossEncoder
from transformers import AutoTokenizer, pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# ΓöÇΓöÇ Load data ΓöÇΓöÇ
train = pd.read_csv('/kaggle/input/competitions/smart-mcq-solver-challenge/train.csv')
LABELS = ['A', 'B', 'C', 'D', 'E']

# ΓöÇΓöÇ Build knowledge base: correct answer text for each training row ΓöÇΓöÇ
print("Creating knowledge base...")
kb = []
for idx, row in train.iterrows():
    correct_letter = row['answer']
    kb.append(str(row[correct_letter]))

# ΓöÇΓöÇ Embed KB and build FAISS index ΓöÇΓöÇ
print("Loading embedding model and creating index...")
model = SentenceTransformer('all-MiniLM-L6-v2')
kb_embeddings = model.encode(kb, show_progress_bar=False)
index = faiss.IndexFlatL2(kb_embeddings.shape[1])
index.add(kb_embeddings)
print(f"Knowledge base created: {len(kb)} documents, index dimension={kb_embeddings.shape[1]}")
```

### Zero-shot classifier setup

```python
# ΓöÇΓöÇ Zero-shot classifier ΓöÇΓöÇ
zs = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# ΓöÇΓöÇ Row 150 setup (used in Q1, Q2, Q5, Q6) ΓöÇΓöÇ
row_150 = train.iloc[150]
prompt_150 = str(row_150['prompt'])
labels_150 = [str(row_150[c]) for c in LABELS]
ans_150 = str(row_150[row_150['answer']])
print(f"Row 150 prompt: {prompt_150[:80]}...")
print(f"Correct answer: option {row_150['answer']} = '{ans_150[:60]}...'")
```

## Q1 ΓÇö Zero-Shot Probability of Correct Option (Row 150)

Run `bart-large-mnli` zero-shot classifier on the prompt for row 150,
passing the 5 options as `candidate_labels`. What probability does the
model assign to the ground-truth correct option?

```python
result = zs(prompt_150, candidate_labels=labels_150)

# Find the probability for the correct answer text
correct_idx = result['labels'].index(ans_150)
q1_answer = round(result['scores'][correct_idx], 3)
print(f"Q1: Probability of correct option = {q1_answer}")
```

## Q2 ΓÇö FAISS Retrieval Rank for True Document (Row 150)

Embed the prompt for row 150 and query FAISS for top-10 nearest
documents. At what rank (1ΓÇô10) does the true correct document
(KB index 150) appear?

```python
query_embedding = model.encode([prompt_150])
distances, retrieved_indices = index.search(query_embedding, k=10)
retrieved_indices = retrieved_indices[0]  # flatten

print(f"Top-10 retrieved KB indices: {retrieved_indices.tolist()}")

# Find rank of the true document (KB index 150)
rank_of_true = list(retrieved_indices).index(150) + 1  # 1-indexed
q2_answer = rank_of_true
print(f"Q2: True document rank = {q2_answer}")
```

## Q3 ΓÇö Cross-Encoder Reranking (Row 150)

Take the top-10 FAISS results, re-rank them using the Cross-Encoder
`ms-marco-MiniLM-L-6-v2`. At what rank does the true document end up?

This demonstrates the two-stage retrieval pipeline:
1. **Bi-Encoder (FAISS)**: fast but imprecise ΓåÆ true doc at rank 10
2. **Cross-Encoder**: slow but precise ΓåÆ true doc at rank 1

```python
cross_encoder = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')

docs_10 = [kb[i] for i in retrieved_indices]
pairs = [[prompt_150, doc] for doc in docs_10]
ce_scores = cross_encoder.predict(pairs)

# Sort by cross-encoder score (descending)
reranked_order = np.argsort(-ce_scores)
reranked_indices = retrieved_indices[reranked_order]

print(f"Cross-encoder scores: {[f'{s:.3f}' for s in ce_scores]}")
print(f"Reranked order: {reranked_indices.tolist()}")

rank_of_true_ce = list(reranked_indices).index(150) + 1
q3_answer = rank_of_true_ce
print(f"\nQ3: True document rank after reranking = {q3_answer}")
```

## Q4 ΓÇö Token Count of RAG String (Row 42)

Retrieve top-5 documents for row 42, concatenate them, and form a
RAG string. How many tokens does `bert-base-uncased` produce
(without truncation)?

```python
row_42 = train.iloc[42]
prompt_42 = str(row_42['prompt'])

q42_embedding = model.encode([prompt_42])
_, idx_42 = index.search(q42_embedding, k=5)
docs_5 = [kb[i] for i in idx_42[0]]

concatenated = " ".join(docs_5)
rag_string = f"Context: {concatenated} Question: {prompt_42}"

bert_tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
tokens = bert_tokenizer(rag_string)  # no truncation

q4_answer = len(tokens['input_ids'])
print(f"RAG string length: {len(rag_string)} chars")
print(f"Q4: Total tokens = {q4_answer}")
```

## Q5 ΓÇö Zero-Shot with RAG Context (Row 150)

Use the **true** document from KB as context (perfect retrieval).
How much does the correct-option probability improve?

```python
true_doc = kb[150]  # the true correct document for row 150
rag_prompt = f"Context: {true_doc} Question: {prompt_150}"

result_rag = zs(rag_prompt, candidate_labels=labels_150)
correct_idx_rag = result_rag['labels'].index(ans_150)
q5_answer = round(result_rag['scores'][correct_idx_rag], 3)

print(f"Without RAG (Q1): {q1_answer}")
print(f"With RAG (Q5):    {q5_answer}")
print(f"Improvement:      {q5_answer - q1_answer:+.3f}")
print(f"\nQ5: Probability with RAG context = {q5_answer}")
```

## Q6 ΓÇö Adversarial RAG (Wrong Context, Row 150)

What happens when we inject **incorrect** context? We force KB index 999
(an unrelated document) as the context. This demonstrates the "Garbage In,
Garbage Out" danger of RAG.

```python
wrong_doc = kb[999]  # completely unrelated document
adversarial_prompt = f"Context: {wrong_doc} Question: {prompt_150}"

result_adv = zs(adversarial_prompt, candidate_labels=labels_150)
correct_idx_adv = result_adv['labels'].index(ans_150)
q6_answer = round(result_adv['scores'][correct_idx_adv], 3)

print(f"Without context (Q1):     {q1_answer}")
print(f"With correct context (Q5): {q5_answer}")
print(f"With WRONG context (Q6):   {q6_answer}")
print(f"\nQ6: Probability with adversarial context = {q6_answer}")
print("ΓåÆ Wrong context degrades performance vs no context at all!")
```

## Q7 ΓÇö Hit Rate (First 100 Rows, k=5)

A "hit" occurs if the correct answer text is found in any of the top-5
retrieved documents. What's the hit rate for rows 0ΓÇô99?

```python
hits = 0
for i in range(100):
    row = train.iloc[i]
    prompt = str(row['prompt'])
    correct_text = str(row[row['answer']])

    q_emb = model.encode([prompt])
    _, top5_idx = index.search(q_emb, k=5)
    retrieved_docs = [kb[j] for j in top5_idx[0]]

    # Check if the correct answer text appears in any retrieved doc
    if any(correct_text in doc for doc in retrieved_docs):
        hits += 1

q7_answer = round(hits / 100 * 100, 1)
print(f"Hits: {hits}/100")
print(f"Q7: Hit Rate = {q7_answer}%")
```

## Q8 ΓÇö Full RAG Pipeline MAP@3 (First 20 Rows)

For each of the first 20 rows:
1. **Retrieve** top-5 docs from FAISS
2. **Rerank** with Cross-Encoder ΓåÆ pick the best document
3. **Augment** prompt with that context
4. **Predict** via zero-shot classifier
5. **Score** with MAP@3

```python
def map_at_3_single(true_label, pred_ranking):
    """MAP@3 for a single sample."""
    for rank, pred in enumerate(pred_ranking[:3]):
        if pred == true_label:
            return 1.0 / (rank + 1)
    return 0.0

scores = []
for i in range(20):
    row = train.iloc[i]
    prompt = str(row['prompt'])
    options = [str(row[c]) for c in LABELS]
    correct = row['answer']

    # 1. Retrieve top-5
    q_emb = model.encode([prompt])
    _, top5_idx = index.search(q_emb, k=5)
    docs = [kb[j] for j in top5_idx[0]]

    # 2. Rerank with Cross-Encoder ΓåÆ select best document
    pairs = [[prompt, doc] for doc in docs]
    ce_scores_i = cross_encoder.predict(pairs)
    best_doc = docs[np.argmax(ce_scores_i)]

    # 3. Augment prompt
    augmented = f"Context: {best_doc} Question: {prompt}"

    # 4. Zero-shot classification
    result = zs(augmented, candidate_labels=options)

    # 5. Map back to letters and score
    # result['labels'] is sorted by probability; map option text ΓåÆ letter
    option_to_letter = {opt: LABELS[j] for j, opt in enumerate(options)}
    pred_letters = [option_to_letter[label] for label in result['labels']]

    score = map_at_3_single(correct, pred_letters)
    scores.append(score)
    print(f"Row {i:2d}: true={correct}, pred_top3={pred_letters[:3]}, MAP@3={score:.3f}")

q8_answer = round(np.mean(scores), 3)
print(f"\nQ8: Average MAP@3 over 20 rows = {q8_answer}")
```
