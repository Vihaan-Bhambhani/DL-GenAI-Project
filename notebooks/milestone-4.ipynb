## Q3 ΓÇö Single-Row MCQ Tokenization

Multiple-choice models expect inputs in the shape: `batch_size x num_choices x sequence_length`. Since each question has five options, every row becomes five tokenized sequences.

Using `bert-base-uncased`, tokenize the five formatted inputs for row index 0 with:
- `padding = "max_length"`
- `truncation = True`
- `max_length = 128`
- `return_tensors = "pt"`

After reshaping for a multiple-choice model, the final `input_ids` tensor has shape: `[1, 5, 128]`

What is the value of the second dimension?

````python
from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained("bert-base-uncased")

# Create 5 formatted inputs for row 0
firsts = [str(row0["prompt"])] * 5
seconds = [str(row0["prompt"]) + " [SEP] " + str(row0[c]) for c in LABELS]

enc = tok(firsts, seconds, padding="max_length", truncation=True,
          max_length=128, return_tensors="pt")

# Reshape: (5, 128) -> (1, 5, 128)
input_ids = enc["input_ids"].unsqueeze(0)  # (1, 5, 128)
q3 = input_ids.shape[1]
print(f"input_ids shape: {input_ids.shape}")
print(f"Q3: Second dimension = {q3}")
````

## Q4 ΓÇö Batch MCQ Tokenization

Tokenize the first 16 rows of `train.csv` as multiple-choice examples. Each row has 5 choices. Each choice is tokenized to length 128. The final `input_ids` tensor has shape: `[16, 5, 128]`

How many total token positions are in this tensor?

````python
rows16 = train.iloc[:16]
all_ids = []
for _, row in rows16.iterrows():
    firsts = [str(row["prompt"])] * 5
    seconds = [str(row["prompt"]) + " [SEP] " + str(row[c]) for c in LABELS]
    enc = tok(firsts, seconds, padding="max_length", truncation=True,
              max_length=128, return_tensors="pt")
    all_ids.append(enc["input_ids"])

batch_ids = torch.stack(all_ids)  # (16, 5, 128)
q4 = batch_ids.numel()
print(f"Batch shape: {batch_ids.shape}")
print(f"Q4: Total token positions = {q4}")
````

## Q5 ΓÇö Multiple-Choice Logits

`AutoModelForMultipleChoice` produces one logit score for each answer option. For this competition, the model outputs five logits corresponding to A, B, C, D, and E.

Load `bert-base-uncased` using `AutoModelForMultipleChoice`. Tokenize row index 0 as 5 choices and pass it through the model. The output logits tensor has shape: `[1, 5]`

How many logits are produced for one question?

````python
from transformers import AutoModelForMultipleChoice

mc_model = AutoModelForMultipleChoice.from_pretrained("bert-base-uncased")

firsts = [str(row0["prompt"])] * 5
seconds = [str(row0["prompt"]) + " [SEP] " + str(row0[c]) for c in LABELS]
enc = tok(firsts, seconds, padding="max_length", truncation=True,
          max_length=128, return_tensors="pt")

mc_input = {k: v.unsqueeze(0) for k, v in enc.items()}  # add batch dim
with torch.no_grad():
    out = mc_model(**mc_input)

q5 = out.logits.shape[1]
print(f"Logits shape: {out.logits.shape}")
print(f"Q5: Number of logits = {q5}")
````

## Q6 ΓÇö Supervised Loss Tensor

For row index 0, pass the tokenized 5-choice input into `AutoModelForMultipleChoice` along with the correct encoded label. The model returns a scalar loss tensor.

How many dimensions does this loss tensor have?

````python
label0 = torch.tensor([label_map[row0["answer"]]])
mc_input["labels"] = label0

out_loss = mc_model(**mc_input)
q6 = out_loss.loss.dim()
print(f"Loss tensor: {out_loss.loss}")
print(f"Loss shape: {out_loss.loss.shape}")
print(f"Q6: Number of dimensions = {q6}")
````

## Q7 ΓÇö LoRA Trainable Parameters

LoRA freezes most of the original model and trains only a small number of adapter parameters. This makes fine-tuning faster and more memory-efficient.

Apply LoRA to the `bert-base-uncased` multiple-choice model using:
- `r = 8`
- `lora_alpha = 16`
- `target_modules = ["query", "value"]`
- `lora_dropout = 0.1`
- `bias = "none"`
- `task_type = TaskType.SEQ_CLS`

Count trainable parameters using:
```python
sum(p.numel() for p in model.parameters() if p.requires_grad)
```

How many parameters are trainable?

````python
!pip install --upgrade torchao

from transformers import AutoModelForMultipleChoice
from peft import LoraConfig, get_peft_model, TaskType

# 1. Load your base model
base_model = AutoModelForMultipleChoice.from_pretrained("bert-base-uncased")

# 2. Use SEQ_CLS for multiple-choice tasks
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["query", "value"],
    lora_dropout=0.1,
    bias="none",
    task_type=TaskType.SEQ_CLS,  # <-- Keep SEQ_CLS here
)

# 3. Create your PEFT model
lora_model = get_peft_model(base_model, lora_config)

# 4. Count and verify your trainable parameters
q7 = sum(p.numel() for p in lora_model.parameters() if p.requires_grad)
lora_model.print_trainable_parameters()
print(f"Q7: Trainable parameters = {q7}")
````

## Q8 ΓÇö Hugging Face Dataset Preparation

Before training, the dataset must be converted into a format that the Hugging Face Trainer can understand: tokenized `input_ids`, `attention_mask`, and numeric labels.

Create a Hugging Face `Dataset` from the first 100 rows of `train.csv`. For each row, create:
- `input_ids` with shape `[5, 128]`
- `attention_mask` with shape `[5, 128]`
- `labels` as the encoded answer label

For the first dataset item, `input_ids` has shape: `[5, 128]`

How many tokenized choices are stored in `input_ids`?

````python
from datasets import Dataset

def tokenize_mcq(row):
    firsts = [str(row["prompt"])] * 5
    seconds = [str(row["prompt"]) + " [SEP] " + str(row[c]) for c in LABELS]
    enc = tok(firsts, seconds, padding="max_length", truncation=True, max_length=128)
    return {
        "input_ids": enc["input_ids"],
        "attention_mask": enc["attention_mask"],
        "labels": label_map[row["answer"]]
    }

rows100 = [tokenize_mcq(train.iloc[i]) for i in range(100)]
hf_ds = Dataset.from_list(rows100)
hf_ds.set_format("torch")

q8 = len(hf_ds[0]["input_ids"])
print(f"First item input_ids shape: {hf_ds[0]['input_ids'].shape}")
print(f"Q8: Number of tokenized choices = {q8}")
````

## Q9 ΓÇö Tiny LoRA Fine-Tuning

Fine-tune a LoRA multiple-choice model on the first 32 rows using Hugging Face Trainer. Use the following settings:
- `max_length = 64`
- `per_device_train_batch_size = 4`
- `gradient_accumulation_steps = 1`
- `max_steps = 4`

What is the final `global_step` reported by the Trainer?

````python
from transformers import Trainer, TrainingArguments

# Build dataset with max_length=64
def tokenize_mcq_64(row):
    firsts = [str(row["prompt"])] * 5
    seconds = [str(row["prompt"]) + " [SEP] " + str(row[c]) for c in LABELS]
    enc = tok(firsts, seconds, padding="max_length", truncation=True, max_length=64)
    return {
        "input_ids": enc["input_ids"],
        "attention_mask": enc["attention_mask"],
        "labels": label_map[row["answer"]]
    }

train_rows = [tokenize_mcq_64(train.iloc[i]) for i in range(32)]
train_ds = Dataset.from_list(train_rows)
train_ds.set_format("torch")

# Fresh LoRA model
base_m = AutoModelForMultipleChoice.from_pretrained("bert-base-uncased")
lora_m = get_peft_model(base_m, lora_config)

args = TrainingArguments(
    output_dir="/tmp/lora_tiny",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=1,
    max_steps=4,
    logging_steps=1,
    save_strategy="no",
    report_to="none",
    remove_unused_columns=False,
)

trainer = Trainer(model=lora_m, args=args, train_dataset=train_ds)
result = trainer.train()

q9 = trainer.state.global_step
print(f"Q9: Final global_step = {q9}")
````

## Q10 ΓÇö Probability Assigned to Option E After Fine-Tuning

Using the fine-tuned LoRA model from Q9, run inference on row index 0 and apply softmax to the logits.

What is the probability assigned to Option E? Round your answer to 4 decimal places.

````python
# Inference on row 0
test_row = tokenize_mcq_64(train.iloc[0])
test_input = {
    "input_ids": torch.tensor(test_row["input_ids"]).unsqueeze(0),
    "attention_mask": torch.tensor(test_row["attention_mask"]).unsqueeze(0),
}

lora_m.eval()
with torch.no_grad():
    logits = lora_m(**test_input).logits

probs = torch.softmax(logits, dim=1)[0]
q10 = round(probs[4].item(), 4)  # Option E = index 4
print(f"Probabilities: {[round(p.item(), 4) for p in probs]}")
print(f"Q10: Probability of Option E = {q10}")
````
